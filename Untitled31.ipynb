# Explicación del Código - Detector de Variable Aleatoria

## Descripción General

Este código simula un sistema de comunicación digital utilizando un detector de variable aleatoria. Los símbolos binarios $D0$ y $D1$ son generados y transmitidos a través de un canal con ruido blanco aditivo (AWGN). El objetivo es detectar correctamente los símbolos transmitidos basándose en la primera muestra de cada símbolo recibido, calculando el umbral óptimo para decidir entre los dos símbolos. A continuación, se detallan los principales pasos y elementos involucrados en el proceso.

## Generación de la Señal

1. **Generación de la forma de onda para cada símbolo:**

    - Para el símbolo $D0$, se genera una señal con forma Gaussiana.
    - Para el símbolo $D1$, se genera una señal con forma invertida de la Gaussiana.

   Las formas de onda de ambos símbolos son generadas en función de una envolvente dependiente del tiempo, controlada por un parámetro de varianza.

    - La señal para el símbolo $D0$ se genera con la varianza:
    $$
    \sigma_0(t) = \sigma_0^{\text{max}} \cdot \exp\left(-\frac{t^2}{2\sigma_{\text{env}}^2}\right)
    $$
    - La señal para el símbolo $D1$ se genera de manera similar pero con la varianza invertida:
    $$
    \sigma_1(t) = \sigma_1^{\text{max}} \cdot \left(1 - \exp\left(-\frac{t^2}{2\sigma_{\text{env}}^2}\right)\right)
    $$

2. **Generación de la trama transmitida:** Los bits aleatorios se asignan a los símbolos $D0$ y $D1$ y se transmiten a través del canal. Cada símbolo es representado como una secuencia de muestras, y la señal transmitida es una secuencia de estos símbolos.

## Paso por el Canal AWGN

La señal transmitida pasa por un canal con ruido blanco aditivo (AWGN), modelado por:

$$
r_x(t) = x(t) + \eta(t)
$$

Donde $x(t)$ es la señal transmitida y $\eta(t)$ es el ruido del canal, el cual es generado como ruido gaussiano con varianza controlada por el parámetro `noise_channel`.

## Detector de Variable Aleatoria

El detector utiliza la primera muestra de cada símbolo recibido, denotada como $y(t_0)$. La variable aleatoria $Y$ es la muestra de la señal recibida en el primer instante de cada símbolo:

$$
Y = r_x(t_0)
$$

Se calculan las estadísticas de $Y$ para los símbolos $D0$ y $D1$:

- Para los símbolos $D0$, la media y la desviación estándar son:
$$
\mu_0 = \mathbb{E}[Y | D0], \quad \sigma_0 = \text{Desviación estándar de } Y | D0
$$

- Para los símbolos $D1$, la media y la desviación estándar son:
$$
\mu_1 = \mathbb{E}[Y | D1], \quad \sigma_1 = \text{Desviación estándar de } Y | D1
$$

### Cálculo del Umbral MAP

El umbral de decisión óptimo, $u_{\text{real}}$, se calcula usando el criterio de máxima probabilidad a posteriori (MAP), que depende de las estadísticas de $Y$ para ambos símbolos. Si las desviaciones estándar son diferentes ($\sigma_0 \neq \sigma_1$), el umbral se calcula mediante la solución de una ecuación cuadrática:

$$
A = \sigma_1^2 - \sigma_0^2, \quad B = 2(\mu_0 \sigma_1^2 - \mu_1 \sigma_0^2)
$$
$$
C = \mu_1^2 \sigma_0^2 - \mu_0^2 \sigma_1^2 + 2 \sigma_0^2 \sigma_1^2 \log\left(\frac{\sigma_1}{\sigma_0}\right)
$$

El discriminante de esta ecuación cuadrática es:
$$
\Delta = B^2 - 4AC
$$

Finalmente, el umbral es:
$$
u_{\text{real}} = \frac{-B \pm \sqrt{\Delta}}{2A}
$$

Si la desviación estándar es igual ($\sigma_0 = \sigma_1$), el umbral se calcula simplemente como el punto medio entre las medias:
$$
u_{\text{real}} = \frac{\mu_0 + \mu_1}{2}
$$

### Decisión

La decisión sobre el símbolo transmitido se toma comparando la variable aleatoria $Y$ con el umbral $u_{\text{real}}$. Si $Y \geq u_{\text{real}}$, se decide que el símbolo transmitido fue $D1$; de lo contrario, se decide que fue $D0$.

La tasa de error de bit (BER) se calcula comparando las decisiones con los símbolos transmitidos. El BER es la fracción de bits detectados incorrectamente:

$$
\text{BER} = \frac{1}{N_{\text{symbols}}} \sum_{i=1}^{N_{\text{symbols}}} \mathbb{1} \left( \hat{b}_i \neq b_i \right)
$$

## Distribuciones y Regiones de Decisión

Se calculan las distribuciones de probabilidad para las muestras $Y | D0$ y $Y | D1$ asumiendo una distribución normal para cada caso:

$$
p(Y | D0) = \frac{1}{\sqrt{2 \pi \sigma_0^2}} \exp\left( -\frac{(Y - \mu_0)^2}{2 \sigma_0^2} \right)
$$

$$
p(Y | D1) = \frac{1}{\sqrt{2 \pi \sigma_1^2}} \exp\left( -\frac{(Y - \mu_1)^2}{2 \sigma_1^2} \right)
$$

Las regiones de decisión se definen en función del umbral $u_{\text{real}}$:

- Para $Y < u_{\text{real}}$, se decide $D0$.
- Para $Y \geq u_{\text{real}}$, se decide $D1$.

## Probabilidad de Error

La probabilidad de error se calcula utilizando la función de error complementaria (erfc), que mide el área bajo la curva de distribución a partir del umbral $u_{\text{real}}$:

$$
P_{\text{error} | D0} = \frac{1}{2} \, \text{erfc} \left( \frac{u_{\text{real}} - \mu_0}{\sigma_0 \sqrt{2}} \right)
$$

$$
P_{\text{error} | D1} = \frac{1}{2} \, \text{erfc} \left( \frac{\mu_1 - u_{\text{real}}}{\sigma_1 \sqrt{2}} \right)
$$

La probabilidad total de error es la media de las probabilidades de error para ambos símbolos:
$$
P_{\text{error}} = \frac{1}{2} \left( P_{\text{error} | D0} + P_{\text{error} | D1} \right)
$$

## Conclusión

Este código demuestra cómo se puede diseñar un detector de variable aleatoria para un sistema de comunicación digital, basándose en las estadísticas de la señal recibida y calculando un umbral de decisión óptimo. La probabilidad de error y la tasa de error de bit (BER) son importantes métricas para evaluar el desempeño del sistema.


```python
import torch
import math
import matplotlib.pyplot as plt
from scipy.special import erfc
plt.style.use("ggplot")

# ==========================================================
# CONFIGURACIÓN TORCH
# ==========================================================
torch.set_default_dtype(torch.float64)
torch.manual_seed(42)

# ==========================================================
# PARÁMETROS DEL SISTEMA
# ==========================================================
Nsymbols = 128
Ns = 200
noise_channel = 1.0
Nview = 8

COLOR0 = "#2E86AB"  # Azul para D0
COLOR1 = "#A23B72"  # Magenta para D1

# ==========================================================
# ETIQUETAS DE SÍMBOLOS EN LAS GRÁFICAS
# ==========================================================
def etiquetar_simbolos(symbols, Ns, ax, y_pos=None):
    if y_pos is None:
        ylim = ax.get_ylim()
        y_pos = float(ylim[1] * 0.85)
    for i, s in enumerate(symbols):
        color = COLOR0 if s == 0 else COLOR1
        ax.text(i*Ns + Ns/3, y_pos, f"D{s}", fontsize=12,
                fontweight="bold", color=color)

# ==========================================================
# GENERACIÓN DE VARIANZA GAUSSIANA / INVERTIDA
# ==========================================================
def generar_varianzas(Ns, sigma0_max=3.0, sigma1_max=3.0):
    t = torch.linspace(-1, 1, Ns)
    sigma_env = 0.35
    g = torch.exp(-t**2 / (2*sigma_env**2))
    sigma0_t = sigma0_max * g              # montaña
    sigma1_t = sigma1_max * (1 - g)        # valle invertido
    return sigma0_t, sigma1_t

# ==========================================================
# GENERAR UN SÍMBOLO SEGÚN SU FORMA (D0 o D1)
# ==========================================================
def generar_simbolo(bit, Ns):
    sigma0_t, sigma1_t = generar_varianzas(Ns)
    ruido = torch.randn(Ns) * (sigma0_t if bit == 0 else sigma1_t)
    return ruido

# ==========================================================
# MOSTRAR UN SÍMBOLO D0 Y UN D1
# ==========================================================
print("🌊 GENERANDO FORMAS DE ONDA...\n")

sim_D0 = generar_simbolo(0, Ns).numpy()
sim_D1 = generar_simbolo(1, Ns).numpy()

fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(14, 4))
ax1.plot(sim_D0, color=COLOR0, linewidth=2)
ax1.set_title("SÍMBOLO D0 — forma Gaussiana ", fontweight='bold', color=COLOR0)
ax1.grid(True)

ax2.plot(sim_D1, color=COLOR1, linewidth=2)
ax2.set_title("SÍMBOLO D1 — forma invertida ", fontweight='bold', color=COLOR1)
ax2.grid(True)
plt.tight_layout()
plt.show()

# ==========================================================
# GENERAR TRAMA COMPLETA EN TORCH
# ==========================================================
print("📡 GENERANDO SEÑAL...\n")

bits = torch.randint(0, 2, (Nsymbols,))
symbols = bits.clone()
tx_signal = torch.zeros(Nsymbols * Ns)

for i, b in enumerate(symbols):
    tx_signal[i*Ns:(i+1)*Ns] = generar_simbolo(int(b), Ns)

# 1. TRAMA TRANSMITIDA
plt.figure(figsize=(14,4))
for i in range(Nview):
    seg = tx_signal[i*Ns:(i+1)*Ns]
    color = COLOR0 if symbols[i] == 0 else COLOR1
    plt.plot(range(i*Ns, (i+1)*Ns), seg, color=color, linewidth=1.5)
ax = plt.gca()
etiquetar_simbolos(symbols[:Nview].numpy(), Ns, ax)
plt.title("1. Trama transmitida (símbolos por forma)", fontweight='bold')
plt.grid(True)
plt.tight_layout()
plt.show()

# ==========================================================
# PASAR POR CANAL AWGN
# ==========================================================
awgn = noise_channel * torch.randn(Nsymbols * Ns)
rx_channel = tx_signal + awgn

# 2. SEÑAL RECIBIDA
plt.figure(figsize=(14,4))
for i in range(Nview):
    seg = rx_channel[i*Ns:(i+1)*Ns]
    color = COLOR0 if symbols[i] == 0 else COLOR1
    plt.plot(range(i*Ns, (i+1)*Ns), seg, color=color, linewidth=1.5)
ax = plt.gca()
etiquetar_simbolos(symbols[:Nview].numpy(), Ns, ax)
plt.title("2. Señal recibida después del canal", fontweight='bold')
plt.grid(True)
plt.tight_layout()
plt.show()

# 3. RUIDO + SEÑAL TRANSMITIDA
plt.figure(figsize=(14,4))
plt.plot(awgn[:Nview*Ns], label="Ruido del canal", color='gray', alpha=0.6, linewidth=1.5)
for i in range(Nview):
    seg = tx_signal[i*Ns:(i+1)*Ns]
    color = COLOR0 if symbols[i] == 0 else COLOR1
    plt.plot(range(i*Ns, (i+1)*Ns), seg, color=color, linewidth=2)
plt.plot([], [], color=COLOR0, linewidth=2, label='Señal TX (D0)')
plt.plot([], [], color=COLOR1, linewidth=2, label='Señal TX (D1)')

ax = plt.gca()
y_pos = float(torch.max(tx_signal[:Nview*Ns]) * 1.1)
etiquetar_simbolos(symbols[:Nview].numpy(), Ns, ax, y_pos=y_pos)
plt.title("3. Ruido del canal + Trama transmitida", fontweight='bold')
plt.grid(True)
plt.legend()
plt.tight_layout()
plt.show()

# ==========================================================
# 🎯 DETECTOR DE VARIABLE ALEATORIA
# ==========================================================
print("🎯 DETECTOR DE VARIABLE ALEATORIA...\n")

# Reorganizar en bloques
rx_matrix = rx_channel.reshape(Nsymbols, Ns)

# ⭐ VARIABLE ALEATORIA: Primera muestra de cada símbolo
y = rx_matrix[:, 0]

# Separar según el símbolo transmitido
y0 = y[symbols == 0]
y1 = y[symbols == 1]

# Estadísticas
m0, s0 = float(y0.mean()), float(y0.std())
m1, s1 = float(y1.mean()), float(y1.std())

print(f"✅ Variable Aleatoria: Y (primera muestra)")
print(f"   Y|D0 → μ={m0:.3f}, σ={s0:.3f}")
print(f"   Y|D1 → μ={m1:.3f}, σ={s1:.3f}\n")

# Umbral MAP
if abs(s1 - s0) < 1e-6:
    u_real = (m0 + m1) / 2
else:
    A = s1*s1 - s0*s0
    B = 2*(m0*s1*s1 - m1*s0*s0)
    C = m1*m1*s0*s0 - m0*m0*s1*s1 + 2*s0*s0*s1*s1*math.log(s1/s0)
    disc = max(B*B - 4*A*C, 0)
    u1 = (-B + math.sqrt(disc)) / (2*A)
    u2 = (-B - math.sqrt(disc)) / (2*A)
    u_real = u1 if min(m0,m1) <= u1 <= max(m0,m1) else u2

# Decisión basada en Y
y_hat = (y >= u_real).long()
BER = float((y_hat != symbols).float().mean())

print(f"   Umbral MAP: {u_real:.4f}")
print(f"   BER: {BER:.5f}\n")

# ==========================================================
# 4. GRÁFICA DE LA VARIABLE ALEATORIA
# ==========================================================
y0_idx = torch.where(symbols == 0)[0]
y1_idx = torch.where(symbols == 1)[0]

plt.figure(figsize=(14, 5))
plt.scatter(y0_idx, y0, c=COLOR0, s=60, alpha=0.7, label=f"Y|D0 (n={len(y0)})")
plt.scatter(y1_idx, y1, c=COLOR1, s=60, alpha=0.7, label=f"Y|D1 (n={len(y1)})")
plt.axhline(u_real, color="red", ls="--", linewidth=2.5,
            label=f"Umbral MAP = {u_real:.3f}")

# Marcar errores
errores = (y_hat != symbols)
if errores.any():
    y_err = y[errores]
    idx_err = torch.where(errores)[0]
    plt.scatter(idx_err, y_err, c="orange", s=150, marker='x',
                linewidths=3, label=f"Errores ({errores.sum()})", zorder=5)

plt.title("4. Variable Aleatoria Y (primera muestra)",
          fontweight='bold', fontsize=13)
plt.xlabel("Índice del símbolo")
plt.ylabel("Valor de Y")
plt.legend()
plt.grid(True)
plt.tight_layout()
plt.show()

# ==========================================================
# 5. DISTRIBUCIONES Y REGIONES DE DECISIÓN
# ==========================================================
print("📊 DISTRIBUCIONES...\n")

xmin = min(m0 - 4*s0, m1 - 4*s1)
xmax = max(m0 + 4*s0, m1 + 4*s1)
xx = torch.linspace(xmin, xmax, 1000)

p0 = torch.exp(-(xx-m0)**2/(2*s0**2)) / (s0*math.sqrt(2*math.pi))
p1 = torch.exp(-(xx-m1)**2/(2*s1**2)) / (s1*math.sqrt(2*math.pi))

fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(15, 5))

# Histogramas + PDFs
ax1.hist(y0.numpy(), bins=30, alpha=0.6, color=COLOR0,
         label=f'Y|D0', density=True, edgecolor='black')
ax1.hist(y1.numpy(), bins=30, alpha=0.6, color=COLOR1,
         label=f'Y|D1', density=True, edgecolor='black')
ax1.plot(xx, p0, color=COLOR0, linewidth=2.5, linestyle='--', alpha=0.9)
ax1.plot(xx, p1, color=COLOR1, linewidth=2.5, linestyle='--', alpha=0.9)
ax1.axvline(u_real, color='red', linewidth=2.5, linestyle='--',
            label=f'Umbral={u_real:.3f}')
ax1.set_xlabel('Y (valor)', fontsize=11)
ax1.set_ylabel('Densidad', fontsize=11)
ax1.set_title('Distribución de la Variable Aleatoria Y', fontweight='bold')
ax1.legend()
ax1.grid(True, alpha=0.3)

# Regiones de decisión
ax2.fill_between(xx[xx < u_real], 0,
                 torch.maximum(p0[xx < u_real], p1[xx < u_real]),
                 alpha=0.4, color=COLOR0, label='Región: decidir D0')
ax2.fill_between(xx[xx >= u_real], 0,
                 torch.maximum(p0[xx >= u_real], p1[xx >= u_real]),
                 alpha=0.4, color=COLOR1, label='Región: decidir D1')
ax2.plot(xx, p0, color=COLOR0, linewidth=2.5, label='p(Y|D0)')
ax2.plot(xx, p1, color=COLOR1, linewidth=2.5, label='p(Y|D1)')
ax2.axvline(u_real, color='red', linewidth=2.5, linestyle='--')
ax2.set_xlabel('Y (valor)', fontsize=11)
ax2.set_ylabel('Densidad', fontsize=11)
ax2.set_title('Regiones de Decisión', fontweight='bold')
ax2.legend()
ax2.grid(True, alpha=0.3)

plt.tight_layout()
plt.show()

# ==========================================================
# 6. PROBABILIDAD DE ERROR
# ==========================================================
print("📉 PROBABILIDAD DE ERROR...\n")

Pe_0 = 0.5 * erfc((u_real - m0) / (s0 * math.sqrt(2)))
Pe_1 = 0.5 * erfc((m1 - u_real) / (s1 * math.sqrt(2)))
Pe = 0.5 * (Pe_0 + Pe_1)

plt.figure(figsize=(13, 6))
plt.plot(xx, p0, color=COLOR0, linewidth=2.5, label="p(Y|D0)")
plt.plot(xx, p1, color=COLOR1, linewidth=2.5, label="p(Y|D1)")
plt.axvline(u_real, color="red", ls="--", linewidth=2.5,
            label=f"Umbral MAP = {u_real:.3f}")

# Áreas de error
x_err0 = xx[xx >= u_real]
p0_err = torch.exp(-(x_err0-m0)**2/(2*s0**2)) / (s0*math.sqrt(2*math.pi))
plt.fill_between(x_err0, 0, p0_err, color=COLOR0, alpha=0.4,
                 label=f"P(error|D0) = {Pe_0:.5f}")

x_err1 = xx[xx < u_real]
p1_err = torch.exp(-(x_err1-m1)**2/(2*s1**2)) / (s1*math.sqrt(2*math.pi))
plt.fill_between(x_err1, 0, p1_err, color=COLOR1, alpha=0.4,
                 label=f"P(error|D1) = {Pe_1:.5f}")

plt.title(f"Probabilidad de Error | Pe = {Pe:.6f} | BER = {BER:.6f}",
          fontsize=13, fontweight='bold')
plt.xlabel("Y (valor)", fontsize=11)
plt.ylabel("Densidad", fontsize=11)
plt.legend(fontsize=10)
plt.grid(True, alpha=0.3)
plt.tight_layout()
plt.show()

# ==========================================================
# 7. COMPARACIÓN TRANSMITIDO VS DETECTADO
# ==========================================================
plt.figure(figsize=(14,5))

indices = torch.arange(Nview)
plt.plot(indices - 0.1, symbols[:Nview], "o-", markersize=12, linewidth=2.5,
         color='red', alpha=0.7, label="Transmitidos", markeredgecolor='darkred', markeredgewidth=2)
plt.plot(indices + 0.1, y_hat[:Nview], "s-", markersize=12, linewidth=2.5,
         color='blue', alpha=0.7, label="Detectados", markeredgecolor='darkblue', markeredgewidth=2)

# Marcar errores
for i in range(Nview):
    if symbols[i] != y_hat[i]:
        plt.plot([i-0.1, i+0.1], [symbols[i], y_hat[i]],
                'x-', color='orange', markersize=15, linewidth=4,
                markeredgewidth=4, alpha=0.9)
        if i == torch.where(symbols[:Nview] != y_hat[:Nview])[0][0]:
            plt.plot([], [], 'x-', color='orange', markersize=15,
                    linewidth=4, label='Error', alpha=0.9)

plt.title("7. Bits transmitidos vs detectados (Variable Aleatoria)",
          fontweight='bold', fontsize=13)
plt.xlabel("Índice del símbolo", fontsize=11)
plt.ylabel("Bit", fontsize=11)
plt.yticks([0, 1])
plt.xticks(range(Nview))
plt.grid(True, alpha=0.4)
plt.legend(fontsize=11)
plt.tight_layout()
plt.show()

print(f"\n{'='*60}")
print(f"RESULTADOS - DETECTOR DE VARIABLE ALEATORIA")
print(f"{'='*60}")
print(f"P(error | D0) = {Pe_0:.6f}")
print(f"P(error | D1) = {Pe_1:.6f}")
print(f"P(error total) = {Pe:.6f}")
print(f"BER medido    = {BER:.6f}")
print(f"{'='*60}")
```

# Explicación del Código - Detector de Señal Aleatoria Basado en Energía

## Descripción General

Este código simula un sistema de comunicación digital utilizando un detector de energía para los símbolos binarios $D0$ y $D1$. Los símbolos son transmitidos a través de un canal con ruido blanco aditivo (AWGN), y el objetivo es detectar correctamente los símbolos basándose en la energía de la señal recibida. El detector calcula un umbral de decisión óptimo que minimiza la probabilidad de error, y la tasa de error de bit (BER) es utilizada como medida de desempeño.

## Generación de la Señal

1. **Generación de la forma de onda para cada símbolo:**

    - Para el símbolo $D0$, se genera una señal con forma Gaussiana.
    - Para el símbolo $D1$, se genera una señal con forma invertida de la Gaussiana.

   Ambas señales son moduladas con una envolvente controlada por una varianza dependiente del tiempo. Las formas de onda de cada símbolo se generan utilizando una función de varianza Gaussiana:

    - Para $D0$:
    $$
    \sigma_0(t) = \sigma_0^{\text{max}} \cdot \exp\left(-\frac{t^2}{2\sigma_{\text{env}}^2}\right)
    $$

    - Para $D1$:
    $$
    \sigma_1(t) = \sigma_1^{\text{max}} \cdot \left(1 - \exp\left(-\frac{t^2}{2\sigma_{\text{env}}^2}\right)\right)
    $$

2. **Generación de la trama transmitida:** Los bits aleatorios generados se asignan a los símbolos $D0$ y $D1$. La señal transmitida es una secuencia de estas formas de onda correspondientes a los símbolos generados.

## Paso por el Canal AWGN

La señal transmitida pasa por un canal ruidoso con ruido blanco aditivo (AWGN). El modelo matemático de la señal recibida es el siguiente:

$$
r_x(t) = x(t) + \eta(t)
$$

Donde:
- $x(t)$ es la señal transmitida,
- $\eta(t)$ es el ruido blanco gaussiano con varianza controlada por el parámetro `noise_channel`.

Este ruido afecta la señal recibida, y se visualiza junto con la señal transmitida para evaluar el impacto del canal.

## Detector de Energía

El detector basado en energía calcula la energía de cada símbolo recibido. La energía de un símbolo se obtiene como la suma de los cuadrados de los valores de la señal en cada instante de tiempo:

$$
E = \sum_t |r(t)|^2
$$

Se calcula la energía para cada símbolo y se compara con un umbral de decisión. La decisión sobre el símbolo recibido se toma en función de la energía medida en cada muestra.

### Cálculo del Umbral de Decisión

El umbral óptimo de decisión se calcula utilizando el valor medio de las energías de ambos símbolos. Si las energías medias de los símbolos son diferentes, el umbral se calcula como el promedio de las medias de energía de ambos símbolos. Si las energías son iguales, el umbral se calcula como el punto medio de estas energías.

$$
u_{\text{real}} = \frac{E_0 + E_1}{2}
$$

### Decisión

La decisión sobre el símbolo transmitido se toma comparando la energía medida con el umbral calculado. Si la energía es mayor que el umbral, se decide que el símbolo transmitido fue $D1$; de lo contrario, se decide que fue $D0$.

## Distribuciones y Regiones de Decisión

Se calculan las distribuciones de energía para ambos símbolos ($D0$ y $D1$), asumiendo que ambas distribuciones siguen una distribución normal. La probabilidad de que la energía sea asociada a un símbolo específico está dada por:

- Para $D0$:
$$
p(E | D0) = \frac{1}{\sqrt{2 \pi \sigma_0^2}} \exp\left( -\frac{(E - \mu_0)^2}{2 \sigma_0^2} \right)
$$

- Para $D1$:
$$
p(E | D1) = \frac{1}{\sqrt{2 \pi \sigma_1^2}} \exp\left( -\frac{(E - \mu_1)^2}{2 \sigma_1^2} \right)
$$

Las regiones de decisión se definen de la siguiente manera:
- Para $E < u_{\text{real}}$, se decide $D0$.
- Para $E \geq u_{\text{real}}$, se decide $D1$.

## Probabilidad de Error

La probabilidad de error se calcula utilizando la función de error complementaria (erfc), que mide la probabilidad de tomar una decisión incorrecta debido al ruido en el canal. La probabilidad de error para cada símbolo se calcula como:

- Para $D0$:
$$
P_{\text{error} | D0} = \frac{1}{2} \, \text{erfc} \left( \frac{u_{\text{real}} - \mu_0}{\sigma_0 \sqrt{2}} \right)
$$

- Para $D1$:
$$
P_{\text{error} | D1} = \frac{1}{2} \, \text{erfc} \left( \frac{\mu_1 - u_{\text{real}}}{\sigma_1 \sqrt{2}} \right)
$$

La probabilidad total de error se calcula promediando ambas probabilidades:

$$
P_{\text{error}} = \frac{1}{2} \left( P_{\text{error} | D0} + P_{\text{error} | D1} \right)
$$

## Conclusión

Este código muestra cómo diseñar un sistema de comunicación digital con detección basada en energía. Al calcular un umbral óptimo basado en las energías de los símbolos transmitidos y recibidos, es posible tomar decisiones sobre el símbolo transmitido. La probabilidad de error y la tasa de error de bit (BER) son evaluadas para medir la efectividad del sistema de detección.


```python
import torch
import math
import matplotlib.pyplot as plt
from scipy.special import erfc
plt.style.use("ggplot")

# ==========================================================
# CONFIGURACIÓN TORCH
# ==========================================================
torch.set_default_dtype(torch.float64)
torch.manual_seed(42)

# ==========================================================
# PARÁMETROS DEL SISTEMA
# ==========================================================
Nsymbols = 128
Ns = 200
noise_channel = 1.0
Nview = 8

COLOR0 = "#2E86AB"
COLOR1 = "#A23B72"

# ==========================================================
# ETIQUETAS DE SÍMBOLOS EN LAS GRÁFICAS
# ==========================================================
def etiquetar_simbolos(symbols, Ns, ax, y_pos=None):
    if y_pos is None:
        ylim = ax.get_ylim()
        y_pos = float(ylim[1] * 0.85)
    for i, s in enumerate(symbols):
        color = COLOR0 if s == 0 else COLOR1
        ax.text(i*Ns + Ns/3, y_pos, f"D{s}", fontsize=12,
                fontweight="bold", color=color)

# ==========================================================
# GENERACIÓN DE VARIANZA GAUSSIANA / INVERTIDA
# ==========================================================
def generar_varianzas(Ns, sigma0_max=3.0, sigma1_max=3.0):
    t = torch.linspace(-1, 1, Ns)
    sigma_env = 0.35
    g = torch.exp(-t**2 / (2*sigma_env**2))
    sigma0_t = sigma0_max * g
    sigma1_t = sigma1_max * (1 - g)
    return sigma0_t, sigma1_t

# ==========================================================
# GENERAR UN SÍMBOLO SEGÚN SU FORMA (D0 o D1)
# ==========================================================
def generar_simbolo(bit, Ns):
    sigma0_t, sigma1_t = generar_varianzas(Ns)
    ruido = torch.randn(Ns) * (sigma0_t if bit == 0 else sigma1_t)
    return ruido

# ==========================================================
# MOSTRAR UN SÍMBOLO D0 Y UN D1
# ==========================================================
print("🌊 GENERANDO FORMAS DE ONDA...\n")

sim_D0 = generar_simbolo(0, Ns).numpy()
sim_D1 = generar_simbolo(1, Ns).numpy()

fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(14, 4))
ax1.plot(sim_D0, color=COLOR0, linewidth=2)
ax1.set_title("SÍMBOLO D0 — forma Gaussiana ", fontweight='bold')
ax1.grid(True)

ax2.plot(sim_D1, color=COLOR1, linewidth=2)
ax2.set_title("SÍMBOLO D1 — forma invertida ", fontweight='bold')
ax2.grid(True)
plt.tight_layout()
plt.show()

# ==========================================================
# GENERAR TRAMA COMPLETA EN TORCH
# ==========================================================
print("📡 GENERANDO SEÑAL...\n")

bits = torch.randint(0, 2, (Nsymbols,))
symbols = bits.clone()
tx_signal = torch.zeros(Nsymbols * Ns)

for i, b in enumerate(symbols):
    tx_signal[i*Ns:(i+1)*Ns] = generar_simbolo(int(b), Ns)

# 1. TRAMA TRANSMITIDA
plt.figure(figsize=(14,4))
for i in range(Nview):
    seg = tx_signal[i*Ns:(i+1)*Ns]
    color = COLOR0 if symbols[i] == 0 else COLOR1
    plt.plot(range(i*Ns, (i+1)*Ns), seg, color=color, linewidth=1.5)
ax = plt.gca()
etiquetar_simbolos(symbols[:Nview].numpy(), Ns, ax)
plt.title("1. Trama transmitida (símbolos por forma)", fontweight='bold')
plt.grid(True)
plt.tight_layout()
plt.show()

# ==========================================================
# PASAR POR CANAL AWGN
# ==========================================================
awgn = noise_channel * torch.randn(Nsymbols * Ns)
rx_channel = tx_signal + awgn

# 2. SEÑAL RECIBIDA
plt.figure(figsize=(14,4))
for i in range(Nview):
    seg = rx_channel[i*Ns:(i+1)*Ns]
    color = COLOR0 if symbols[i] == 0 else COLOR1
    plt.plot(range(i*Ns, (i+1)*Ns), seg, color=color, linewidth=1.5)
ax = plt.gca()
etiquetar_simbolos(symbols[:Nview].numpy(), Ns, ax)
plt.title("2. Señal recibida después del canal", fontweight='bold')
plt.grid(True)
plt.tight_layout()
plt.show()

# 3. RUIDO + SEÑAL TRANSMITIDA
plt.figure(figsize=(14,4))
plt.plot(awgn[:Nview*Ns], label="Ruido del canal", color='gray', alpha=0.6, linewidth=1.5)
for i in range(Nview):
    seg = tx_signal[i*Ns:(i+1)*Ns]
    color = COLOR0 if symbols[i] == 0 else COLOR1
    plt.plot(range(i*Ns, (i+1)*Ns), seg, color=color, linewidth=2)
plt.plot([], [], color=COLOR0, linewidth=2, label='Señal TX (D0)')
plt.plot([], [], color=COLOR1, linewidth=2, label='Señal TX (D1)')

ax = plt.gca()
y_pos = float(torch.max(tx_signal[:Nview*Ns]) * 1.1)
etiquetar_simbolos(symbols[:Nview].numpy(), Ns, ax, y_pos=y_pos)
plt.title("3. Ruido del canal + Trama transmitida", fontweight='bold')
plt.grid(True)
plt.legend()
plt.tight_layout()
plt.show()

# ==========================================================
# 🎯 DETECTOR DE SEÑAL ALEATORIA
# ==========================================================
print("🎯 DETECTOR DE SEÑAL ALEATORIA...\n")

# Reorganizar en bloques (matriz de señales)
rx_matrix = rx_channel.reshape(Nsymbols, Ns)

# ⭐ ESTADÍSTICO: ENERGÍA DE CADA SEÑAL COMPLETA
energias = (rx_matrix ** 2).sum(dim=1)

# Separar energías según el símbolo transmitido
E0 = energias[symbols == 0]
E1 = energias[symbols == 1]

# Estadísticas de las energías
m0, s0 = float(E0.mean()), float(E0.std())
m1, s1 = float(E1.mean()), float(E1.std())

print(f"✅ Estadístico: E = Σ r²(t)")
print(f"   E(D0) → μ={m0:.1f}, σ={s0:.1f}")
print(f"   E(D1) → μ={m1:.1f}, σ={s1:.1f}\n")

# Umbral óptimo (punto medio para caso simétrico)
thr_energia = (m0 + m1) / 2

# Decisión basada en energía
bits_hat = (energias >= thr_energia).long()
BER = float((bits_hat != symbols).float().mean())

print(f"   Umbral: {thr_energia:.1f}")
print(f"   BER: {BER:.5f}\n")

# ==========================================================
# 3. GRÁFICA DEL ESTADÍSTICO
# ==========================================================
E0_idx = torch.where(symbols == 0)[0]
E1_idx = torch.where(symbols == 1)[0]

plt.figure(figsize=(14, 5))
plt.scatter(E0_idx, E0, c=COLOR0, s=60, alpha=0.7, label=f"E|D0 (n={len(E0)})")
plt.scatter(E1_idx, E1, c=COLOR1, s=60, alpha=0.7, label=f"E|D1 (n={len(E1)})")
plt.axhline(thr_energia, color="red", ls="--", linewidth=2.5,
            label=f"Umbral = {thr_energia:.1f}")

# Marcar errores
errores = (bits_hat != symbols)
if errores.any():
    E_err = energias[errores]
    idx_err = torch.where(errores)[0]
    plt.scatter(idx_err, E_err, c="orange", s=150, marker='x',
                linewidths=3, label=f"Errores ({errores.sum()})", zorder=5)

plt.title("3. Estadístico de Energía por Símbolo (Señal Aleatoria)",
          fontweight='bold', fontsize=13)
plt.xlabel("Índice del símbolo")
plt.ylabel("Energía E = Σ r²(t)")
plt.legend()
plt.grid(True)
plt.tight_layout()
plt.show()

# ==========================================================
# 4. DISTRIBUCIONES Y REGIONES DE DECISIÓN
# ==========================================================
print("📊 DISTRIBUCIONES...\n")

x = torch.linspace(max(0, m0-4*s0), m1+4*s1, 1000)
p0 = torch.exp(-(x-m0)**2/(2*s0**2)) / (s0*math.sqrt(2*math.pi))
p1 = torch.exp(-(x-m1)**2/(2*s1**2)) / (s1*math.sqrt(2*math.pi))

fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(15, 5))

# Histogramas + PDFs
ax1.hist(E0.numpy(), bins=30, alpha=0.6, color=COLOR0,
         label=f'E|D0', density=True, edgecolor='black')
ax1.hist(E1.numpy(), bins=30, alpha=0.6, color=COLOR1,
         label=f'E|D1', density=True, edgecolor='black')
ax1.plot(x, p0, color=COLOR0, linewidth=2.5, linestyle='--', alpha=0.9)
ax1.plot(x, p1, color=COLOR1, linewidth=2.5, linestyle='--', alpha=0.9)
ax1.axvline(thr_energia, color='red', linewidth=2.5, linestyle='--',
            label=f'Umbral={thr_energia:.1f}')
ax1.set_xlabel('E (Energía)', fontsize=11)
ax1.set_ylabel('Densidad', fontsize=11)
ax1.set_title('Distribución del Estadístico E', fontweight='bold')
ax1.legend()
ax1.grid(True, alpha=0.3)

# Regiones de decisión
ax2.fill_between(x[x < thr_energia], 0,
                 torch.maximum(p0[x < thr_energia], p1[x < thr_energia]),
                 alpha=0.4, color=COLOR0, label='Región: decidir D0')
ax2.fill_between(x[x >= thr_energia], 0,
                 torch.maximum(p0[x >= thr_energia], p1[x >= thr_energia]),
                 alpha=0.4, color=COLOR1, label='Región: decidir D1')
ax2.plot(x, p0, color=COLOR0, linewidth=2.5, label='p(E|D0)')
ax2.plot(x, p1, color=COLOR1, linewidth=2.5, label='p(E|D1)')
ax2.axvline(thr_energia, color='red', linewidth=2.5, linestyle='--')
ax2.set_xlabel('E (Energía)', fontsize=11)
ax2.set_ylabel('Densidad', fontsize=11)
ax2.set_title('Regiones de Decisión', fontweight='bold')
ax2.legend()
ax2.grid(True, alpha=0.3)

plt.tight_layout()
plt.show()

# ==========================================================
# 5. PROBABILIDAD DE ERROR
# ==========================================================
print("📉 PROBABILIDAD DE ERROR...\n")

Pe_0 = 0.5 * erfc((thr_energia - m0) / (s0 * math.sqrt(2)))
Pe_1 = 0.5 * erfc((m1 - thr_energia) / (s1 * math.sqrt(2)))
Pe = 0.5 * (Pe_0 + Pe_1)

plt.figure(figsize=(13, 6))
plt.plot(x, p0, color=COLOR0, linewidth=2.5, label="p(E|D0)")
plt.plot(x, p1, color=COLOR1, linewidth=2.5, label="p(E|D1)")
plt.axvline(thr_energia, color="red", ls="--", linewidth=2.5,
            label=f"Umbral = {thr_energia:.1f}")

# Áreas de error
x_err0 = x[x >= thr_energia]
p0_err = torch.exp(-(x_err0-m0)**2/(2*s0**2)) / (s0*math.sqrt(2*math.pi))
plt.fill_between(x_err0, 0, p0_err, color=COLOR0, alpha=0.4,
                 label=f"P(error|D0) = {Pe_0:.5f}")

x_err1 = x[x < thr_energia]
p1_err = torch.exp(-(x_err1-m1)**2/(2*s1**2)) / (s1*math.sqrt(2*math.pi))
plt.fill_between(x_err1, 0, p1_err, color=COLOR1, alpha=0.4,
                 label=f"P(error|D1) = {Pe_1:.5f}")

plt.title(f"Probabilidad de Error | Pe = {Pe:.6f} | BER = {BER:.6f}",
          fontsize=13, fontweight='bold')
plt.xlabel("E (Energía)", fontsize=11)
plt.ylabel("Densidad", fontsize=11)
plt.legend(fontsize=10)
plt.grid(True, alpha=0.3)
plt.tight_layout()
plt.show()

# ==========================================================
# 6. COMPARACIÓN TRANSMITIDO VS DETECTADO
# ==========================================================
plt.figure(figsize=(14,5))

# Transmitidos en rojo con offset
indices = torch.arange(Nview)
plt.plot(indices - 0.1, symbols[:Nview], "o-", markersize=12, linewidth=2.5,
         color='red', alpha=0.7, label="Transmitidos", markeredgecolor='darkred', markeredgewidth=2)

# Detectados en azul con offset
plt.plot(indices + 0.1, bits_hat[:Nview], "s-", markersize=12, linewidth=2.5,
         color='blue', alpha=0.7, label="Detectados", markeredgecolor='darkblue', markeredgewidth=2)

# Marcar errores
for i in range(Nview):
    if symbols[i] != bits_hat[i]:
        plt.plot([i-0.1, i+0.1], [symbols[i], bits_hat[i]],
                'x-', color='orange', markersize=15, linewidth=4,
                markeredgewidth=4, alpha=0.9)
        if i == torch.where(symbols[:Nview] != bits_hat[:Nview])[0][0]:
            plt.plot([], [], 'x-', color='orange', markersize=15,
                    linewidth=4, label='Error', alpha=0.9)

plt.title("6. Bits transmitidos vs detectados (Señal Aleatoria)",
          fontweight='bold', fontsize=13)
plt.xlabel("Índice del símbolo", fontsize=11)
plt.ylabel("Bit", fontsize=11)
plt.yticks([0, 1])
plt.xticks(range(Nview))
plt.grid(True, alpha=0.4)
plt.legend(fontsize=11)
plt.tight_layout()
plt.show()

print(f"\n{'='*60}")
print(f"RESULTADOS - DETECTOR DE SEÑAL ALEATORIA")
print(f"{'='*60}")
print(f"P(error | D0) = {Pe_0:.6f}")
print(f"P(error | D1) = {Pe_1:.6f}")
print(f"P(error total) = {Pe:.6f}")
print(f"BER medido    = {BER:.6f}")
print(f"{'='*60}")
```
